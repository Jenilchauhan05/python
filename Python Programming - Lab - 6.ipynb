### 01) WAP to find sum of all the elements in a List.

```python
n = int(input("enter the length"))
new = []
sum = 0

for i in range(n) :
    inp = int(input("enter the numbers"))
    new.append(inp)
print(new)

for i in range(n) :
    sum += new[i]
    
print("sum = ", sum)
```

### 04) WAP to interchange first and last elements in a list.

```python
n = int(input("enter the length"))
new = []

for i in range(n) :
    inp = int(input("enter the numbers"))
    new.append(inp)
print("old list : ", new)

temp = new[len(new) - 1]
new[len(new) - 1] = new[0]
new[0] = temp

print("new list : ", new)
```

### 16) WAP to create a new list (fruit whose name starts with 'b') from the list of fruits given by user.

```python
n = int(input("enter the length"))
new = []

for i in range(n) :
    inp = input("enter the elements")
    new.append(inp)
print("original list : ", new)

character = input("enter the first character of string")
new_lst = []
for i in new :
    if(i.startswith(character)) :
        new_lst.append(i)

print("new list : ", new_lst)
```

### 17) WAP to create a list of common elements from given two lists.

```python
n1 = int(input("enter the length of l1"))
new1 = []

for i in range(n1) :
    inp = int(input("enter the elements"))
    new1.append(inp)

n2 = int(input("enter the length of l2"))
new2 = []

for i in range(n2) :
    inp = int(input("enter the elements"))
    new2.append(inp)

print("list 1 : ", new1)
print("list 2 : ", new2)

new_lst = []
for i in new1 :
    if(i in new2 and i not in new_lst) :
        new_lst.append(i)

print("comman element list : ", new_lst)
```
