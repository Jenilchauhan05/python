### 04) WAP to find the longest word(s) in a file.

```python
with open("data.txt", "r") as file :
    words = file.read().split()
    max_len = len(max(words, key=len))
    longest = [word for word in words if len(word) == max_len]

print("longest word : ", longest)
```

### 05) WAP to count the no. of lines, words and characters in a given text file.

```python
f = open("data.txt", "r")
lines = f.readlines()

linecount = len(lines)
wordcount = sum(len(line.split()) for line in lines)
charactercount = sum(len(line) for line in lines)
print(linecount, wordcount, charactercount)
f.close()
```

### 08) WAP to create an UDF named frequency to count occurances of the specific word in a given text file.

```python
def frequency(name, word) :
    f = open(name, "r")
    data = f.read().split()
    print(data.count(word))

frequency("data.txt", "to")
```

### 09) WAP to get the score of five subjects from the user, store them in a file. Fetch those marks and find the highest score.

```python
f = open("marks.txt", "w")

for i in range(5):
    marks = int(input("Enter marks: "))
    f.write(str(marks) + "\n")

f.close()

f = open("marks.txt", "r")

data = f.read().split("\n")
nums = []

for x in data:
    if x != "":
        nums.append(int(x))

print("Highest Marks:", max(nums))

f.close()
```

### 10) WAP to write first 100 prime numbers to a file named primenumbers.txt 
(Note: each number should be in new line)

```python
f = open("primenumbers.txt", "w")
count = 0      
num = 2        

while count < 100:
    isprime = True

    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            isprime = False
            break

    if isprime:
        f.write(str(num) + "\n")
        count += 1

    num += 1

f.close()
```
