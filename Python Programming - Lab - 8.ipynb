### 01) WAP to find their union, intersection, and difference of the given two sets.

### 05) WAP to Merge two dictionaries. If a key exists in both dictionaries, add their values.

**Sample Input:**
```
Dict1 = {1: 10, 2: 20, 3: 30}
Dict2 = {2: 40, 3: 50, 4: 60}

```
**Sample Output:**
```
{1: 10, 2: 60, 3: 80, 4: 60}
```

````python
d1 = {}
d2 = {}

pairs1 = input("enter key, value pairs in d1").split(",")
pairs2 = input("enter key, value pairs in d2").split(",")

for i in pairs1 :
    k, v = i.split(":")
    d1[int(k)] = int(v)

for i in pairs2 :
    k, v = i.split(":")
    d2[int(k)] = int(v)

res = d1.copy()

for key, value in d2.items() :
    if key in res :
        res[key] += value
    else :
        res[key] = value

print(res)
````

### 06) WAP to Sort a dictionary in ascending order of its values.
Hint: in sorted(), use udf for key parameter


**Sample Input:**
```
{'Math': 70, 'Physics': 85, 'Chemistry': 60}

```
**Sample Output:**
```
{'Chemistry': 60, 'Math': 70, 'Physics': 85}
```

````python
d1 = {}
p1 = input("enter key, value pairs in d1").split(",")

for i in p1 :
    k, v = i.split(":")
    d1[k] = v

marks = dict(sorted(d1.items(), key = lambda x : x[1]))
print(marks)
````

### 07) WAP to find the keys having maximum and minimum values in a dictionary.

**Sample Input:**
```
{'A': 50, 'B': 90, 'C': 30}
```
**Sample Output:**
```
Max Value Key: B
Min Value Key: C

```

````python
d1 = {}
p1 = input("Enter key:value pairs: ").split(",")

for i in p1:
    k, v = i.split(":")
    d1[k] = int(v)

max_val = float('-inf')
min_val = float('inf')
max_key = ""
min_key = ""

for k, v in d1.items():
    if v > max_val:
        max_val = v
        max_key = k
    if v < min_val:
        min_val = v
        min_key = k

print("Max value key:", max_key)
print("Min value key:", min_key)
````

### 11) WAP to find the student with the highest total marks, given student names and their marks.

```
{'Amit': [70, 80, 90], 'Riya': [85, 75, 80], 'Neha': [60, 70, 65]}

```

### 12) WAP to extract keys whose values are even numbers.

**Sample Input:**
```
{1: 11, 2: 20, 3: 15, 4: 40}
```

```
[2, 4]
```

````python
d1 = {}
l1 = []
p1 = input("Enter key:value pairs: ").split(",")

for i in p1:
    k, v = i.split(":")
    d1[int(k)] = v

for k, v in d1.items() :
    if (k % 2 == 0) :
        l1.append(k)

print(l1)
````

### 13) WAP to Find common keys between two dictionaries.

**Sample Input:**
```
Dict1 = {1: 'A', 2: 'B', 3: 'C'}
Dict2 = {2: 'X', 3: 'Y', 4: 'Z'}

```
**Sample Output:**
```
[2, 3]
```

````python
d1 = {}
d2 = {}

p1 = input("enter key:value pairs in d1 : ").split(",")
p2 = input("enter key:value pairs in d2 : ").split(",")

for i in p1:
    k, v = i.split(":")
    d1[int(k.strip())] = v.strip().strip("'")

for i in p2:
    k, v = i.split(":")
    d2[int(k.strip())] = v.strip().strip("'")

l1 = [k for k in d1 if k in d2]

print(l1)
````

### 15) WAP to create a dictionary using dictionary comprehension from two lists containing student names and marks. Include only those students in the dictionary whose marks are greater than or equal to 75.

**Sample Input:**
```
students = ['Amit', 'Riya', 'Neha', 'Karan']
marks = [78, 72, 85, 60]
```
**Sample Output:**
```
{'Amit': 78, 'Neha': 85}
```

````python
students = ['Amit', 'Riya', 'Neha', 'Karan']
marks = [78, 72, 85, 60]

d1 = {k : v for k, v in zip(students, marks) if v >= 75}
print(d1)
````
