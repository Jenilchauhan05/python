### 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
def isprime(n) :
    for i in range(2, int(n**0.5)+1) :
        if (n % i == 0) :
            return 0

    return 1

isprime(27)
```

### 04) Write a function that returns the list of Prime numbers between given two numbers.

### 05) Write a function that returns True if the given string is Palindrome or False otherwise.

```python
def ispalindrome(n) :
    rev = n[::-1]
    if (rev == n) :
        return True
    else :
        return False

ispalindrome('racecar')
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
def fibo(n) :
    if n == 0 :
        return 0
    elif n == 1 :
        return 1
    else :
        return fibo(n-1) + fibo(n-2)

num = int(input("enter the term"))
fibo(num)
```

### 10) Write a function to get the sum of the scores ending with zero.
#### Example : scores = [200, 456, 300, 100, 234, 678] 
#### Ans = 200 + 300 + 100 = 600

```python
def total_zero(l1) :
    total = 0
    for i in l1 :
        if i % 10 == 0 :
            total += i

    return total

scores = [200, 456, 300, 100, 234, 678]
total_zero(scores)
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
def invert(d1) :
    new = {}
    for k, v in d1.items() :
        new[v] = k

    return new

d = {'a': 10, 'b':20, 'c':30, 'd':40}
invert(d)
```

### 12) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
def total_case(s) :
    upper = lower = 0
    for i in s :
        if i.isupper() :
            upper += 1
        else :
            lower += 1

    print("no_upper = ", upper, "no_lower = ", lower)

s1 = input("enter the string")
total_case(s1)
```

### 16) Write udfs to call the functions with following types of arguments:
1. Positional Arguments
2. Keyword Arguments
3. Default Arguments
4. Variable Legngth Positional(*args) & variable length Keyword Arguments (**kwargs)
5. Keyword-Only & Positional Only Arguments

```python
def stu(name, age) :
    print(age, name)

stu("jenil", 20) #positional

def stu(name, age) :
    print(age, name)

stu(name = "jenil", age = 20) #keyword

def stu(name, age = 20) :
    print(age, name)

stu("jenil") #default

def total(*n) :
    print(n)

total(1, 2, 3, 4, 5) # Variable Legngth Positional(*args)

def stu(**data) :
    print(data)

stu(name = "jenil", age = 20, sem = 4, college = "du") # variable length Keyword Arguments (**kwargs)

def stu(name, age, /) :
    print(age, name)

stu("jenil", 20) # Positional Only Arguments

def stu(*, name, age) :
    print(age, name)

stu(name = "jenil", age = 20) # keyword Only Arguments
```
