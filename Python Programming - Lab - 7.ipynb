### 02) WAP to find Maximum and Minimum K elements in a given tuple.

T = (7, 2, 9, 4, 1, 5)

### 10) WAP to test if the given tuple is distinct or not.
**Sample Input:**
```
tuple1 = (1, 2, 3, 4)
tuple2 = (1, 2, 2, 3)
```
**Sample Output:**
```
(1, 2, 3, 4) is distinct? -> True
(1, 2, 2, 3) is distinct? -> False
```

````python
inp = tuple(map(int, input("enter the numbers").split()))
ans = []
is_distinct = True

for i in inp :
    if i not in ans :
        ans.append(i)
    else :
        is_distinct = False
        break

ans = tuple(ans)
print(inp, "is distinct? :", is_distinct)
````

### 14) WAP to create (index, value) pairs for prime numbers.

**Sample Input:**
```
T = (4, 7, 9, 11, 15)
```
**Sample Output:**
```
((1, 7), (3, 11))
```

````python
inp = tuple(map(int, input("enter the numbers").split()))
ans = []

for i in range(len(inp)) :
    n = inp[i]
    if n > 1 :
        is_prime = True
        for j in range(2, n) :
            if n % j == 0 :
                is_prime = False
                break
        if is_prime :
            ans.append((i, n))

ans = tuple(ans)
print(ans)
````

### 15) WAP to split a tuple into even-index and odd-index tuples.

**Sample Input:**
```
T = ('a', 'b', 'c', 'd', 'e')
```

**Sample Input:**
```
even = ('a', 'c', 'e')
odd = ('b', 'd')
```


````python
inp = tuple(map(int, input("enter the numbers").split()))
even = []
odd = []

for i in range(len(inp)) :
    if i % 2 == 0 :
        odd.append(inp[i])
    else :
        even.append(inp[i])

even = tuple(even)
odd = tuple(odd)
print(even, "even")
print(odd, "odd")
````

### 16) WAP to compute column-wise sum from a tuple of lists.

**Sample Input:**
```
T = ([1, 2, 3], [4, 5, 6], [7, 8, 9])
```

**Sample Output:**
```
12 15 18

```

````python
inp = [tuple(map(int,i.split(','))) for i in input("Enter numbers separated by space: ").split()]

for i in range(len(inp)) :
    sum = 0
    for j in range(len(inp[0])) :
        sum += inp[j][i]
    print(sum)
````
