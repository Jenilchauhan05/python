### 05) WAP to generate 3 random integers between 100 and 999 which is divisible by 5.

```python
count = 0
l1 = []

while count < 3 :
    n = random.randint(100, 999)

    if n % 5 == 0 :
        l1.append(n)
        count += 1

print("list = ", l1)
```

### 06) WAP to generate 100 random lottery tickets and pick two lucky tickets from it and announce them as Winner and Runner up respectively.

```python
ticket = random.sample(range(1000, 9999), k=100)

winner = random.sample(ticket, k=2)

print("winner = ", winner[0])
print("runner up ", winner[1])
```

### 08) Subtract a week (7 days) from a given date in Python.

```python
from datetime import datetime, timedelta

inp = datetime(2026, 3, 3)

new_date = inp - timedelta(days=7)

print("Original date:", inp.strftime("%Y-%m-%d"))
print("new date:", new_date.strftime("%Y-%m-%d"))
```

### 09) WAP to Calculate number of days between two given dates.

```python
date1_str = input("Enter the first date: ")
date2_str = input("Enter the second date: ")

date1 = datetime.strptime(date1_str, "%Y-%m-%d")
date2 = datetime.strptime(date2_str, "%Y-%m-%d")

difference = abs((date2 - date1).days)

print(f"Number of days between {date1_str} and {date2_str} is: {difference} days")
```
