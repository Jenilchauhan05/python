### 1) Character Category Counter
**Problem:** Count uppercase, lowercase, digits, and special characters in a string.

**Sample Inputs & Outputs:**
- Input: `Python3@Lab` → Uppercase: 2, Lowercase: 7, Digits: 1, Special: 1
- Input: `HELLO123` → Uppercase: 5, Lowercase: 0, Digits: 3, Special: 0
- Input: `hi@#9` → Uppercase: 0, Lowercase: 2, Digits: 1, Special: 2

```python
str = input("enter the string")
upp = low = dig = spe = 0
for i in str :
    if (i.islower()) :
        low += 1
    elif (i.isupper()) :
        upp += 1
    elif (i.isdigit()) :
        dig += 1
    else :
        spe += 1
print("uppercase = ", upp, "lowercase = ", low, "digits = ", dig, "special = ", spe)
```

### 2) Reverse Each Word

### 9) String Compression
- `aaabbcccc` → `a3b2c4`
- `xxxyyyz` → `x3y3z1`
- `abcd` → `a1b1c1d1`

```python
s = input("Enter the string: ")
compressed = ""
count = 1

for i in range(len(s)):
    if i < len(s) - 1 and s[i] == s[i + 1]:
        count += 1
    else:
        compressed += s[i] + str(count)
        count = 1

print(compressed)
```

### 10) Toggle Case (without using str.swapcase())
- `PyThOn` → `pYtHoN`
- `HELlo` → `helLO`
- `Python` → `pYTHON`

```python
str = input("enter the string")

for i in str :
    if (i.islower()) :
        str = str.replace(i, i.upper())
    else :
        str = str.replace(i, i.lower())

print(str)
```

### 15) Custom Split
- `apple,banana,grape`
- `10|20|30`

```python
s = input("enter the string")
sep = input("enter the seprator")
word = ""

for i in s :
    if i == sep :
        print(word)
        word = ""
    else :
        word += i

print(word)
```
