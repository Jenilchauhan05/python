### 02) Create a class named Bank_Account with Account_No, User_Name, Email,Account_Type and Account_Balance data members. Also create a method GetAccountDetails() and DisplayAccountDetails(). Create main method to demonstrate the Bank_Account class.

### 03) WAP to create Circle class with area and perimeter function to find area and perimeter of circle.

```python
class Circle :
    def __init__(self, r) :
        self.r = r

    def area(self) :
        return 3.14 * self.r * self.r

    def perimeter(self) :
        return 2 * 3.14 * self.r

def main() :
    rad = float(input("enter radius"))
    c = Circle(rad)

    print("area : ", c.area())
    print("perimeter : ", c.perimeter())

main()
```

### 05) Create a bank account class with methods to deposit, withdraw, and check balance.

```python
class BankAccount :
    def __init__(self, balance) :
        self.balance = balance

    def deposit(self, amount) :
        self.balance = self.balance + amount
        print("deposited amount : ", amount)

    def withdraw(self, amount) :
        if amount <= self.balance :
            self.balance = self.balance - amount
            print("withdrawn amount : ", amount)
        else :
            print("insufficient balance")

    def check_balance(self) :
        print("current balance : ", self.balance)

def main() :
    acc = BankAccount(7000)

    acc.check_balance()

    acc.deposit(2000)
    acc.check_balance()

    acc.withdraw(5000)
    acc.check_balance()

main()
```

### 06) Create a class for managing inventory that includes attributes such as item name, price, quantity, and methods to add, remove, and update items.

```python
class Inventory :
    def __init__(self, name, price, quantity) :
        self.name = name
        self.price = price
        self.quantity = quantity

    def add_items(self, qty) :
        self.quantity = self.quantity + qty
        print(qty, " items added")

    def remove_items(self, qty) :
        if qty <= self.quantity :
            self.quantity = self.quantity - qty
            print(qty, " items removed")
        else :
            print("insufficient items")

    def update_items(self, name, price, quantity) :
        self.name = name
        self.price = price
        self.quantity = quantity
        print("details updated")

    def display(self) :
        print("inventory details")
        print("Item Name : ", self.name)
        print("Price : ", self.price)
        print("quantity : ", self.quantity)

def main() :
    item = Inventory("laptop", 50000, 10)
    item.display()

    item.add_items(5)
    item.display()

    item.remove_items(5)
    item.display()

    item.update_items("mobile", 10000, 5)
    item.display()

main()
```

### 09) Define Time class with hour and minute as data member. Also define addition method to add two time objects.

```python
class Time :
    def __init__(self, hour, min) :
        self.hour = hour
        self.min = min

    def add(self, t2) :
        total_hour = self.hour + t2.hour
        total_min = self.min + t2.min

        if total_min >= 60 :
            total_hour += total_min // 60
            total_min = total_min % 60

        print("added time ", total_hour, "hour ", total_min, "min")

def main() :

    t1 = Time(2, 45)
    t2 = Time(3, 30)

    t1.add(t2)

main()
```
