### Practical 1: Count Lines, Words and Characters

**Problem Definition:**  
Write a Python program to read text from a file and count the total number of lines, words, and characters. The result should be written to an output file.


```python
f = open("data.txt", "r")
lines = f.readlines()

linecount = str(len(lines))
wordcount = str(sum(len(line.split()) for line in lines))
charactercount = str(sum(len(line) for line in lines))
print(linecount, wordcount, charactercount)

f1 = open("output.txt", "w")
f1.write("lines = " + linecount + "\n")
f1.write("words = " + wordcount + "\n")
f1.write("characters = " + charactercount + "\n")
f1.close()

print("replacement done")
```

**Problem Definition:**

```python
f = open("data1.txt", "r+")

def result(marks) :
    total = 0
    data = [int(i) for i in marks]
    for i in data :
        total += i
        
    return total / len(data)


for line in f :
    data = line.split()
    roll = data[0]
    name = data[1]
    marks = list(data[2:])
    print(data[2:])
    print(result(marks))

f1 = open("output.txt", "a")
f1.write(str(result(marks)))
f1.close()
```


### Practical 4: Maximum and Minimum using Tuple

**Problem Definition:**  
Write a Python program to read numbers from a file, store them in a tuple, and find the maximum and minimum values. Write the result to an output file.


```python
f = open("marks.txt", "r")
data = f.read()

numbers = data.split()

t = tuple(int(x) for x in numbers)

maximum = max(t)
minimum = min(t)

f2 = open("output.txt", "w")
f2.write("Numbers in tuple: " + str(t) + "\n")
f2.write("Maximum value: " + str(maximum) + "\n")
f2.write("Minimum value: " + str(minimum))

f.close()
f2.close()
```


### Practical 5: Remove Duplicate Words using Set

**Problem Definition:**  
Write a Python program to read words from a file and remove duplicate words using a set. Store the unique words in an output file.


```python
f = open("data1.txt", "r")
data = f.read()

words = data.split()

unique_words = set(words)

f2 = open("data.txt", "w")

for word in unique_words:
    f2.write(word + " ")

f.close()
f2.close()
```


### Practical 6: Word Frequency using Dictionary

**Problem Definition:**  
Write a Python program to count the frequency of each word in a text file using a dictionary and write the result to an output file.


```python
f = open("copy.txt", "r")
data = f.read()

words = data.split()

freq = {}

for word in words:
    if word in freq:
        freq[word] = freq[word] + 1
    else:
        freq[word] = 1

f2 = open("data.txt", "w")

for key in freq:
    f2.write(key + " : " + str(freq[key]) + "\n")

f.close()
f2.close()
```


### Practical 7: Student Result and Grade Calculation

**Problem Definition:**  
Write a Python program to read roll number, name, and marks of students from a file. Use a user-defined function to determine grade (A/B/C/Fail) and write the student result to an output file.

**Grade Criteria:**  
- Marks ≥ 75 : Grade A  
- Marks ≥ 60 : Grade B  
- Marks ≥ 40 : Grade C  
- Marks < 40 : Fail


```python
def calculate_grade(marks):
    if marks >= 75:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 40:
        return "C"
    else:
        return "Fail"

f = open("merge.txt", "r")

f2 = open("new1.txt", "w")

for line in f:
    data = line.split()
    
    roll = data[0]
    name = data[1]
    marks = int(data[2])

    grade = calculate_grade(marks)

    f2.write(roll + " " + name + " " + str(marks) + " Grade: " + grade + "\n")

f.close()
f2.close()
```
