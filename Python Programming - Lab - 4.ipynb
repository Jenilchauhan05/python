### 10) WAP to print GCD of given two numbers.

```python
a = int(input("enter the a"))
b = int(input("enter the b"))
i = gcd = 1
while (i != a+1 and i != b+1) :
    if (a % i == 0 and b % i == 0) :
        gcd = i
    i += 1
print(gcd)
```

### 12) WAP to find whether the given number is Prime or not.

```python
sum = 0
n = int(input("enter the n"))
for i in range (2, (n//2)+1) :
    if (n % i == 0) :
        sum += 1
        break
    if (sum > 0) :
        print("n is not prime number")
    else:
        print("n is prime")
```

### 15) WAP to check whether the given number is an Armstrong Number or not.

```python
n = int(input("enter the n"))
sum = 0
temp = n
dig = len(str(n))
while n > 0 :
    num = n % 10
    sum += pow(num, dig)
    n = n // 10
if (sum == temp) :
    print("armstrong")
else :
    print("not")
```

### 16) WAP to print all the perfect numbers between 1 to n.

```python
n = int(input("enter the n"))
for i in range(1, n+1) :
    sum = 0
    for j in range(1, i) :
        if (i % j == 0) :
            sum += j
    if (i == sum) :
        print(i)
```
