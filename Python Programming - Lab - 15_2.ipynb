### 10) Calculate area of a ractangle using object as an argument to a method.

```python
class Ractangle :
    def __init__(self, b, h) :
        self.b = b
        self.h = h

    def area(self) :
        return self.b * self.h

def main() :
    breath = int(input("enter the breath"))
    height = int(input("enter the height"))

    r = Ractangle(breath, height)
    print("area of ractangle : ", r.area())

main()
```

### 11) Calculate the area of a square. 
### Include a Constructor, a method to calculate area named area() and a method named output() that prints the output and is invoked by area().

```python
class Square :
    def __init__(self, l) :
        self.l = l

    def output(self, area) :
        print("area of aquare : ", area)

    def area(self) :
        a = self.l * self.l
        self.output(a)

def main() :
    len = int(input("enter length of square"))

    s = Square(len)
    s.area()

main()
```

### 12) Calculate the area of a rectangle. 
### Include a Constructor, a method to calculate area named area() and a method named output() that prints the output and is invoked by area().
### Also define a class method that compares the two sides of reactangle. An object is instantiated only if the two sides are different; otherwise a message should be displayed : THIS IS SQUARE.

```python
class Rectangle :
    def __init__(self, b, h) :
        self.b = b
        self.h = h

    def output(self, area) :
        print("area of rectangle : ", area)

    def area(self) :
        if self.b != self.h :
            a = self.b * self.h
            self.output(a)
        else :
            print("this is square")

def main() :
    breath = int(input("enter the breath"))
    height = int(input("enter the height"))

    r = Rectangle(breath, height)
    r.area()

main()
```

### 14) Create a class Profit that has a method named getProfit that accepts profit from the user.
### Create a class Loss that has a method named getLoss that accepts loss from the user.
### Create a class BalanceSheet that inherits from both classes Profit and Loss and calculates the balanace. It has two methods getBalance() and printBalance().

```python
class Profit :
    def getProfit(self) :
        self.profit = int(input("enter the profit"))

class Loss :
    def getLoss(self) :
        self.loss = int(input("enter the loss"))

class BalanceSheet(Profit, Loss) :
    def getBalance(self) :
        self.balance = self.profit - self.loss

    def printBalance(self) :
        print("profit : ", self.profit)
        print("loss : ", self.loss)
        print("balance : ", self.balance)

def main() :
    b = BalanceSheet()

    b.getProfit()
    b.getLoss()

    b.getBalance()
    b.printBalance()

main()
```

### 15) WAP to demonstrate all types of inheritance.

```python
class GrandParents :
    def showGrandParents(self) :
        print("grand parents class")

class Father(GrandParents) :
    def showFather(self) :
        print("father class")

class Mother(GrandParents) :
    def showMother(self) :
        print("mother class")

class Child(Father, Mother) :
    def showChild(self) :
        print("child class")

c = Child()

c.showGrandParents()
c.showFather()
c.showMother()
c.showChild()
```

### 16) Create a Person class with a constructor that takes two arguments name and age. 
### Create a child class Employee that inherits from Person and adds a new attribute salary. 
### Override the __init__ method in Employee to call the parent class's __init__ method using the super() and then initialize the salary attribute.

```python
class Person :
    def __init__(self, name, age) :
        self.name = name
        self.age = age

class Employee(Person) :
    def __init__(self, name, age, salary) :
        super().__init__(name, age)
        self.salary = salary

    def display(self) :
        print("name : ", self.name)
        print("age : ", self.age)
        print("salary : ", self.salary)

def main() :
    emp = Employee("jenil", 20, 50000)
    emp.display()

main()
```

### 17) Create a Shape class with a draw method that is not implemented. 
### Create three child classes Rectangle, Circle, and Triangle that implement the draw method with their respective drawing behaviors. 
### Create a list of Shape objects that includes one instance of each child class, and then iterate through the list and call the draw method on each object.

```python
class Shape :
    def draw(self) :
        pass

class Rectangle(Shape) :
    def draw(self) :
        print("drawing rectangle")

class Circle(Shape) :
    def draw(self) :
        print("drawing circle")

class Triangle(Shape) :
    def draw(self) :
        print("drawing triangle")

def main() :
    Shapes = [Rectangle(), Circle(), Triangle()]

    for shape in Shapes :
        shape.draw()

main()
```
